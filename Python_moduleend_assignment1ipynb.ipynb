```python
#discount & reward functions

#calling the discount fun

def calculate_discount(bill_amount,vip_member):
  if vip_member == True:
    discounted_total = float(bill_amount * 0.85)
  else:
    discounted_total = float(bill_amount * 0.95)
  return discounted_total

#get the bill amount & check vipmember
Amount = int(input("Enter the bill amount"))
#check_vipmem=bool(input("is vip_member"))
check_vipmem=input("Is VIP member (True/False): ").strip().lower() == "true"

total_amount = calculate_discount(Amount,check_vipmem)

print(f"Bill Amount is: {Amount}")
print(f"Is vip member: {check_vipmem}")
print(f"Revised Amount is: {total_amount}")

```

```python
#Recursion

def tier_bonus(level):
  if (level==1):
    return 50
  else:
    return 50 + tier_bonus(level-1)

Level=int(input("enter the level "))
bonus=tier_bonus(Level)

print(f"no of level is: {Level}")
print(f"bonus points are: {bonus}")
```

# Section5 OOPS

```python
class StoreBranch:
    company_name = "Urban Retail"   # Class Variable

    def __init__(self, branch_id, daily_revenue):
        self.branch_id = branch_id
        self.__daily_revenue = daily_revenue   # Private Variable # encapsulation

    def get_revenue(self):
        return self.__daily_revenue


# Child Class
class FlagshipStore(StoreBranch):

    def __init__(self, branch_id, daily_revenue, vip_lounge_access):
        super().__init__(branch_id, daily_revenue)
        self.vip_lounge_access = vip_lounge_access

    # Method Overriding (Polymorphism)
    def get_revenue(self):
        return super().get_revenue() + 500.0


# Instantiation
store = FlagshipStore(200, 18000.0, True)

# Print final calculated revenue
print("Branch ID:", store.branch_id)
print("Company Name:", StoreBranch.company_name)
print("VIP Lounge Access:", store.vip_lounge_access)
print("Final Revenue:", store.get_revenue())
```
